```python
from datasets import load_dataset

print("Loading data...")
dataset=load_dataset("HuggingFaceH4/ultrachat_200k")

def is_coding_chat(example):
    prompt = example['prompt'].lower()
    return 'python' in prompt or 'code' in prompt

coding_chats=dataset['train_sft'].filter(is_coding_chat)
print(f"We found {len(coding_chats)} coding chats!")
```

```python
import ollama

print("Tutor is ready!Type 'quit' to stop.")
example_chat=coding_chats[0]['messages']

print("Tutor is ready! Type 'quit' to stop.")

while True:
    user_message=input("You: ")
    if user_message.lower()=='quit':
        print("Goodbye!")
        break
    chat_history=[
        {'role':'system','content':'You are a helpful expert AI Coding Tutor. Be short, clear, and always use code examples.'}
    ]
    for msg in example_chat:
        chat_history.append({'role':msg['role'],'content':msg['content']})

    chat_history.append({'role': 'user', 'content': user_message})
    response=ollama.chat(model='llama3', messages=chat_history)
    print("Tutor:", response['message']['content'])
    print("-" * 30)
```
